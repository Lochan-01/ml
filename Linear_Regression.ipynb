```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
```

```python
df= pd.read_csv('/content/Salary Data.csv')
df
```

```python
df.info()
```

```python
#to check the missing value
df.isnull().sum()
print(df.isnull().sum())
```

```python
df.duplicated().sum()
print(df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
print(df.duplicated().sum())
```

```python
# fill numerical values for numerical columns with mean
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
df['Years of Experience']=df['Years of Experience'].fillna(df['Years of Experience'].mean())

#fill mode for object datatype
df['Gender']=df['Gender'].fillna(df['Gender'].mode()[0])
df['Job Title']=df['Job Title'].fillna(df['Job Title'].mode()[0])
df['Education Level']=df['Education Level'].fillna(df['Education Level'].mode()[0])
```

```python
df.isnull().sum()
```

```python
#Encoding
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Education Level']=le.fit_transform(df['Education Level'])

#1-hot encoding
df= pd.get_dummies(df,columns=['Job Title'],drop_first=True)

#mapping
mapping={'Male':1,'Female':0}
df['Gender']=df['Gender'].map(mapping)
```

```python
#we have to change the float parameters
df['Age'] = df['Age'].astype(int)
df['Salary'] = df['Salary'].astype(int)
df['Years of Experience'] = df['Years of Experience'].astype(int)
```
